```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def show_comparison(original, blur3, blur5, blur7):
  plt.figure(figsize=(12,10))
  plt.subplot(2,2,1)
  plt.imshow(original,cmap='gray')
  plt.title("Original Grayscale Image")
  plt.axis('off')

  #3x3 kernal
  plt.subplot(2,2,2)
  plt.imshow(blur3,cmap='gray')
  plt.title("Average Filter(3x3)")
  plt.axis('off')

  #5x5 kernal
  plt.subplot(2,2,3)
  plt.imshow(blur5,cmap='gray')
  plt.title("Average Filter(5x5)")
  plt.axis('off')

  #7x7 kernal
  plt.subplot(2,2,4)
  plt.imshow(blur7,cmap='gray')
  plt.title("Average Filter(7x7)")
  plt.axis('off')

  plt.tight_layout()
  plt.show()

# Load the image and apply filters
img = cv2.imread('/content/new york.jpg',0)
if img is None:
  print("Error: Image Not Found")
else:
  kernel_3x3=np.ones((3,3),np.float32)/9
  kernel_5x5=np.ones((5,5),np.float32)/25
  kernel_7x7=np.ones((7,7),np.float32)/49

  blur_3x3=cv2.filter2D(img,-1,kernel_3x3)
  blur_5x5=cv2.filter2D(img,-1,kernel_5x5)
  blur_7x7=cv2.filter2D(img,-1,kernel_7x7)

  show_comparison(img,blur_3x3,blur_5x5,blur_7x7)
```

```python

img2 = cv2.imread('/content/Noisy.jpg', 0)
if img2 is None:
  print("Error: Image 2 Not Found")
else:
  blur2_3x3 = cv2.filter2D(img2, -1, kernel_3x3)
  blur2_5x5 = cv2.filter2D(img2, -1, kernel_5x5)
  blur2_7x7 = cv2.filter2D(img2, -1, kernel_7x7)

  show_comparison(img2, blur2_3x3, blur2_5x5, blur2_7x7)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("/content/BANTI.jpg", 0)
if img is None:
    print("Error: Image not found or could not be loaded.")
else:
    kernel = np.ones((3, 3)) * (1/9)
    k_s = 3
    m, n = img.shape
    filtered_img = np.zeros((m, n))
    p = 0
    for i in range(0, m):
        for j in range(0, n):
            if (i + k_s <= m) and (j + k_s <= n):
                filtered_img[i][j] = np.sum(
                    kernel * img[i:i+k_s, j:j+k_s]
                )
    filtered_img = np.uint8(filtered_img)
    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.subplot(1, 2, 2)
    plt.imshow(filtered_img, cmap='gray')
    plt.title("Filtered Image")
    plt.show()
```
