```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def show_comparison(orginal,sobel,prewitt,canny):
  plt.figure(figsize=(12,10))
  #Orgianl Image
  plt.subplot(2,2,1)
  plt.imshow(orginal,cmap='gray')
  plt.title('Original GrayScale Image')
  plt.axis('off')
  #Sobel EDGE Detection
  plt.subplot(2,2,2)
  plt.imshow(sobel,cmap='gray')
  plt.title('Sobel Edge Detection')
  plt.axis('off')
  #Prewitt EDGE Detection
  plt.subplot(2,2,3)
  plt.imshow(prewitt,cmap='gray')
  plt.title('Prewitt Edge Detection')
  plt.axis('off')
  #canny EDGE Detection
  plt.subplot(2,2,4)
  plt.imshow(canny,cmap='gray')
  plt.title('Canny Edge Detection')
  plt.axis('off')

  plt.tight_layout()
  plt.show()

#Read an Image
img = cv2.imread('/content/NEWYORK.jpg',0)
if img is None:
  print('Error:image not found')
else:
  #Sobel Operation
  sobelx = cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
  sobely = cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
  sobel_combined = cv2.bitwise_or(sobelx,sobely)
  #Prewitt Operation
  kernelx = np.array([[1,1,1],[0,0,0],[-1,-1,-1]])
  kernely = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
  prewittx = cv2.filter2D(img,cv2.CV_64F,kernelx)
  prewitty = cv2.filter2D(img,cv2.CV_64F,kernely)
  prewitt_combined = cv2.bitwise_or(prewittx,prewitty)
  #Canny Operation
  canny_edges = cv2.Canny(img,100,200)
  #Display Results
  show_comparison(img,np.uint8(np.absolute(sobel_combined)),np.uint8(np.absolute(prewitt_combined)),canny_edges)
```
