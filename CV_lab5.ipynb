```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img=cv2.imread('/content/RE_CGT650.jpg',cv2.IMREAD_GRAYSCALE)
if img is None:
  print('error loading Image')
else:
  plt.figure(figsize=(12,10))
  #Display Orginal Image
  plt.subplot(3,3,1)
  plt.imshow(img,cmap='gray')
  plt.title('Original Image')
  plt.axis('off')
for i in range(8):
  bit_plane=(img>>i)&1
  vis_plane=bit_plane*255
  plt.subplot(3,3,i+2)
  plt.imshow(vis_plane,cmap='gray')
  plt.title(f'Bit Plane {i}')
  plt.axis('off')
plt.tight_layout()
plt.show()
```
